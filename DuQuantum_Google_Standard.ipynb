## 1. Challenge Problem: Dicke State Preparation

Consider an $N$-qubit, weight-$n$ Dicke state:
$$|D^{(N)}_n\rangle = \begin{pmatrix} N \\ n \end{pmatrix}^{-1/2} \sum_k \mathcal{P}_k \left( |0\rangle^{\otimes(N-n)} \otimes |1\rangle^{\otimes n}\right)$$

For $n=1$, this is the $W$ state:
$$|W\rangle = \frac{1}{\sqrt N} \left( |100\ldots 0 \rangle + |010\ldots 0 \rangle + \ldots + |00\ldots 01 \rangle \right)$$

```python
# Generate synthetic benchmark noise environment
env = sample_benchmark_environment()

# Test state preparation for N=4, n=1
circuit, qubits = create_dicke_state(
    num_qubits=4,
    num_excitations=1,
    cz_pauli_error_rate=env["cz_pauli_error_rate"],
    sq_pauli_error_rate=env["sq_pauli_error_rate"],
    zero_errors=env["zero_errors"],
    one_errors=env["one_errors"],
)

print(f"Generated circuit with {len(circuit)} moments on {len(qubits)} qubits:")
print(qubits)
test_circuit(circuit, 1, qubits)
print("Circuit fidelity test passed!")
```
